# 02 · Loading .mat files

A .mat file holds named variables. `stk.load` finds the x axis and the spectra by their shapes, so the names do not matter. It prints every variable it found and what it chose.

```python
import matplotlib.pyplot as plt
import spectrum_toolkit as stk
```

## Load without any settings

```python
c = stk.load('../data/wt_raw.mat')   # W is found as x, wt as 550 spectra
c
```

```python
fig, ax = plt.subplots(figsize=(9, 3.5))
for sp in list(c)[:10]:          # first 10 of the 550 spectra
    ax.plot(sp.x, sp.y, lw=1)
ax.set_xlabel(c.x_unit); ax.set_ylabel(c.y_unit); ax.set_title('wt_raw.mat, first 10 spectra')
plt.show()
```

## Choose the variables by name

If the guess is wrong, name them. A misspelt name gives an error listing the real ones.

```python
c = stk.load('../data/wt_raw.mat', x='W', y='wt')
```
